# De 70 mil a 225 mil en 5 años

Proyección de **DCA mensual** para un retiro parcial de **10 mil USD al año**. Ejecuta las celdas en orden.


## 1. Supuestos


```python
from pathlib import Path
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, matplotlib.ticker as mticker
from PIL import Image
START, ANNUAL, YEARS, TARGET = 70_000.0, 23_000.0, 5, 225_000.0
MONTHLY = ANNUAL / 12
RATES = {"solo_aportes": 0.0, "r8": 0.08, "r10": 0.10}
print(f"Aporte mensual {MONTHLY:,.2f} | solo ahorro {START + ANNUAL*YEARS:,.0f}")
```

## 2. Simular los tres caminos


```python
def path(rate, months=YEARS*12):
    r = (1+rate)**(1/12)-1
    v = np.empty(months+1); v[0]=START
    for m in range(1, months+1):
        v[m] = v[m-1]*(1+r)+MONTHLY
    return v
data = {k: path(r) for k,r in RATES.items()}
pd.DataFrame({"escenario":["0%","8%","10%"],"final":[data["solo_aportes"][-1], data["r8"][-1], data["r10"][-1]]}).round(0)
```

Solo aportes = 185k (no alcanza). 8% = 243k. 10% = 259k.
